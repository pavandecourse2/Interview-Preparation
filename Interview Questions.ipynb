```python
%sql
CREATE TABLE Sales(
    sales_id INT,
    sales_date DATE,
    amount INT
);

insert into Sales values
(1,'2025-09-01',1000),
(2,'2025-09-02',2000),
(3,'2025-09-03',1500),
(4,'2025-09-04',3000)
```

```python
%sql
select * from sales;
```

```python
%sql
select *,sum(amount) over(order by sales_date) as running_total
from sales;
```

```python
display(" ***seond highest salary*** ");
df1= spark.sql("""
    with cte (
select *,row_number() over(order by amount desc) rn
from sales)
select * from cte where rn = 2;

""")

display(df1)

display(" *** Third highest salary *** ");
df2= spark.sql(
    """
with cte (
select *,row_number() over(order by amount desc) rn
from sales)
select * from cte where rn = 3;
"""
)

display(df2)

```

```python
nums=[1,2,3,2,4,5,1,6]
d={}
a=[]
for i in nums:
    if i in d:
        d[i]+= 1
    else:
        d[i] = 1
for key,values in d.items(): #The items() method returns a view object that displays a list of dictionary's (key, value) tuple pairs.
    if values > 1:
        a.append(key)        
print(a)        
```
