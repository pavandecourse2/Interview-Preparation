```python
def is_prime(n):
    if n <= 1:
        return False
    
    # Check numbers from 2 up to the square root of n for efficiency
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False  # Found a divisor, definitely not prime
            
    return True  # Looped completely without finding a divisor, it is prime

# Testing the function
num = int(input("Enter a number to check (or type 'q' to quit): "))
#num = 10
if is_prime(num):
    print(num, "is a prime number")
else:
    print(num, "is not a prime number")          
```

```python
#Two strings are anagrams if they use the same charcaters with the same frequency ,but possible in a different orders
#'listen' and 'silent' both are true
# 'hello' and 'world' both are false
#s1 = 'silent'
#s2 = 'listen'
#sorted(s1) == sorted(s2)


def is_anagram(s1, s2):
    # If lengths are different, they cannot be anagrams
    if len(s1) != len(s2):
        return False
        
    char_count = {}
    
    # Count up for the first string
    for char in s1:
        char_count[char] = char_count.get(char, 0) + 1
    display(char)    
    # Count down for the second string
    for char in s2:
        if char not in char_count:
            return False
        char_count[char] -= 1
        if char_count[char] == 0:
            del char_count[char]
            
    # If the dictionary is empty, all characters matched perfectly
    return len(char_count) == 0

s1 = 'silent'
s2 = 'listen'
print(is_anagram(s1, s2))  # Outputs: True
```

```python
%sql

with cte as(
select *,
       lag(price) over(partition by product_id order by price_date) as prev_price
from product_prices
) ,cte1 as(
select product_id,
       price,
       price_date,
       
    case
       when prev_price  is null then null
       when price > prev_price then 'increase'
       when price < prev_price then 'descrease'
       else 'no change' 
    end    status
from cte      
) 

select product_id 
from cte1
where status is not null and status <> 'no change'
group by product_id
having count(distinct status) = 1 and max(status) = 'increase'
 
```
