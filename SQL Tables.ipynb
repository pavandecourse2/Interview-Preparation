```python
%sql
CREATE TABLE product_prices (
    product_id VARCHAR(20) NOT NULL,
    price DECIMAL(10,2) NOT NULL,
    price_date DATE NOT NULL
);
```

```python
%sql
-- ===================================================================
-- TREND 1: CONTINUOUS INCREASE FOR 3 MONTHS (Jan < Feb < Mar)
-- ===================================================================
INSERT INTO product_prices (product_id, price, price_date) VALUES 
('p_101', 500.00, '2026-01-15'),  -- Jan base price
('p_101', 550.00, '2026-02-15'),  -- Feb increase (+50)
('p_101', 620.00, '2026-03-15'),  -- Mar increase (+70)

('p_102', 1200.00, '2026-01-10'), -- Jan base price
('p_102', 1250.00, '2026-02-12'), -- Feb increase (+50)
('p_102', 1400.00, '2026-03-11') -- Mar increase (+150)

```

```python
%sql
-- ===================================================================
-- TREND 2: FLUCTUATING (2-Month Increase, then 1-Month Decrease)
-- ===================================================================
INSERT INTO product_prices (product_id, price, price_date) VALUES 
('p_201', 100.00, '2026-01-20'),  -- Jan base price
('p_201', 130.00, '2026-02-20'),  -- Feb INCREASE (+30)
('p_201', 95.00,  '2026-03-20'),  -- Mar DECREASE (-35)

('p_202', 80.00,  '2026-01-05'),  -- Jan base price
('p_202', 95.00,  '2026-02-05'),  -- Feb INCREASE (+15)
('p_202', 70.00,  '2026-03-05')  -- Mar DECREASE (-25)

 
```

```python
%sql
 
-- ===================================================================
-- TREND 3: CONTROL GROUP (Flat or Unchanged Prices)
-- ===================================================================
INSERT INTO product_prices (product_id, price, price_date) VALUES 
('p_301', 45.00,  '2026-01-18'),  -- Jan
('p_301', 45.00,  '2026-02-18'),  -- Feb
('p_301', 45.00,  '2026-03-18');  -- Mar
```
