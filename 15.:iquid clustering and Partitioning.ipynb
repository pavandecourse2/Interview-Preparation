```python
 %sql
 CREATE OR REPLACE TABLE training.delta_demo.sales_partitioned (
    order_id        STRING,
      order_date      DATE,
     country         STRING,
      category        STRING,
      amount          DOUBLE
  )
  USING DELTA
  PARTITIONED BY (country, category);
 
```
