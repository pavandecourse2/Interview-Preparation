```python
%sql
create or replace table training.jobs_demo.transaction_summary as
select country,count(*) as number_of_transactions from (
select * from training.jobs_demo.sales_customers customers join training.jobs_demo.sales_transactions transactions on customers.customerID = transactions.customerID
 ) group by country
```
