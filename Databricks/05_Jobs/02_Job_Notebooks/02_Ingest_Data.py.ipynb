## Validation for Limit

```python
row_limit_text = dbutils.widgets.get("limits").strip()

if not row_limit_text.isdigit():
    raise ValueError("row_limit must be a whole number")

row_limit = int(row_limit_text)

if row_limit <= 0:
    raise ValueError("row_limit must be greater than 0")

print(f"no of rows {row_limit}")
```

## Ingest table from sample schema to our own schema

```python
try:
    query = f"""
        CREATE OR REPLACE TABLE training.jobs_demo.`{table_name}`
        AS
        SELECT *
        FROM samples.bakehouse.`{table_name}`
        LIMIT {row_limit}
    """

    spark.sql(query)

    df = spark.table(
        f"training.jobs_demo.{table_name}"
    )

    display(df)

    print(
        f"SUCCESS: Copied {row_limit} rows from "
        f"samples.bakehouse.{table_name} to "
        f"training.jobs_demo.{table_name}"
    )

except Exception as e:
    print(f"Error message is: {e}")
    raise   
```
