```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
import pyspark.sql.functions as F
```

```python
df = spark.read.table("samples.bakehouse.sales_customers")
display(df)
```

```python
df.write.mode("overwrite").format("csv").save("/Volumes/test_catalog/test_schema/myloga")
```

```python
# Create Unity Catalog objects only if they do not exist
spark.sql("CREATE CATALOG IF NOT EXISTS test_catalog")
spark.sql("CREATE SCHEMA IF NOT EXISTS test_catalog.test_schema")
spark.sql("CREATE VOLUME IF NOT EXISTS test_catalog.test_schema.myloga")

# Save DataFrame as CSV files inside the volume
(
    df.write
      .mode("overwrite")
      .option("header", "true")
      .csv("/Volumes/test_catalog/test_schema/myloga/employee_csv")
)
```
