```python
from pyspark.sql import Row

data_employee = [
    Row(employee_id=1, name='Alice', city='New York'),
    Row(employee_id=2, name='Bob', city='Los Angeles'),
    Row(employee_id=3, name='Charlie', city='Chicago'),
    Row(employee_id=4, name='David', city='Houston'),
    Row(employee_id=5, name='Eva', city='Phoenix')
]

df_employee = spark.createDataFrame(data_employee)

# Employee info
df_employee.createOrReplaceTempView('employee')

display(df_employee)
```

```python
%sql
create table if not exists test_catalog.test_schema.target_employee_tb1
select *, current_date as created_date
,current_date as updated_date
from employee
 
```

```python
%sql
merge into test_catalog.test_schema.target_employee_tb1  target
using employee  source
on target.employee_id = source.employee_id

when matched then
update set 
 
target.name = source.name,
target.city = source.city,
target.updated_date = current_date()

when not matched
then insert  (employee_id, name, city, created_date, updated_date)
values (source.employee_id, source.name, source.city, current_date(), current_date())
```
